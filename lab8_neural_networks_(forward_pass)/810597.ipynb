Now you will write the function that computes the forward pass. 
* I provide here a structure that you can follow for your function, but again, feel free to modify it as you see fit.
* Use the sigmoid function as the activation of the hidden layer.
* Don't forget about the biases!
* *It is up to you to think what should be the activation for the output layer.*

```python
def sigmoid_activation(z):
    # your code goes here
    return 1/(1 + np.exp(-z))

def add_bias(X:np.ndarray):
    return np.c_[X, np.ones((X.shape[0], 1))]
```

```python
def two_layer_perceptron(X: np.ndarray, act_hidden, dim_input, dim_hidden, dim_output, act_output=None):
    """
    Implements the forward pass of a two-layer fully connected perceptron.
    
    Parameters
    ----------
    X : a 2-dimensional array
        the input data
    act_hidden : function
        the activation function to be used for the hidden layer
    act_output : function
        the activation function to be used for the output layet
    dim_input : int
        the dimensionality of the input layer
    dim_hidden : int
        the dimensionality of the hidden layer
    dim_output : int
        the dimensionality of the output layer
    Returns
    -------
    y_pred : float
        the output of the computation of the forward pass of the network
    """

    # hidden layer
    X_bb = add_bias(X) # bb = by bias
    W_1 = np.random.randn(dim_input + 1, dim_hidden) # plus 1 for bias
    Z1 = act_hidden(X_bb.dot(W_1))

    # saida
    Z1_bb = add_bias(Z1)
    W_2 = np.random.randn(dim_hidden + 1, dim_output)
    Z2_wa = Z1_bb.dot(W_2)

    if act_output != None:
        Z2 = act_output(Z2_wa)
        return Z2

    return Z2_wa
    
```

Calculate the RMSE of the forward pass. 

```python
y_pred = two_layer_perceptron(X_train, sigmoid_activation, X_train.shape[1], 20, 1)
y_pred
```

```python
# your code goes here
def SSE(yt: np.ndarray, yp: np.ndarray):
    return np.sum(0.5*((yt - yp)**2))

err = SSE(X_train, y_pred)
err
```

Write a function that calculates the forward pass for this multi-class classification problem.
* You will use the sigmoid activation function for the hidden layer.
* For the output layer you will have to write the softmax activation function (you can check the slides)
* __Note:__ you can easily re-use the function that you coded for Part I if you do a simple modification and also include an input argument for the activation of the output layer.

```python
# your code goes here:
def softmax_activation(z):
    
    # Garantir Estabilidade
    z_shift = z - np.max(z, axis=1, keepdims=True)

    z_exp = np.exp(z_shift)

    # calcular a soma das exponenciais por linha
    soma_exp = np.sum(z_exp, axis=1, keepdims=True)

    y_pred = z_exp / soma_exp
    return y_pred
```

```python
# your code goes here: 
y_predI = two_layer_perceptron(X_trainI, sigmoid_activation, X_trainI.shape[1], 32, 10, softmax_activation)
```

Lastly, calculate the error of this forward pass using the cross-entropy loss.

```python
# your code goes here:
def cross_entropy(y_true, y_pred):

    err = -np.sum(y_true * np.log(y_pred + 1e-16) + (1 - y_true)*np.log(1 - y_pred + 1e-16))
    return err

def cross_entropy_categorical(y_true, y_pred, target_is_vector=True):
    y_pred = np.clip(y_pred, 1e-16, 1.0)


    if target_is_vector:
        y_pred_c = y_pred[np.arange(y_pred.shape[0]), y_true]
        err = -np.log(y_pred_c)
    else:
        err = -y_true * np.log(y_pred)

    return np.sum(err)

err = cross_entropy_categorical(y_trainI[:, None], y_predI)
err
```
