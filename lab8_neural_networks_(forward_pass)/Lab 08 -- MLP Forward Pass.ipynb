```python
def sigmoid_activation(z):
    # your code goes here
    return 1/(1 + np.exp(-z))
```

```python
def two_layer_perceptron(X: np.ndarray, activation, dim_input, dim_hidden, dim_output):
    """
    Implements the forward pass of a two-layer fully connected perceptron.
    
    Parameters
    ----------
    X : a 2-dimensional array
        the input data
    activation : function
        the activation function to be used for the hidden layer
    dim_input : int
        the dimensionality of the input layer
    dim_hidden : int
        the dimensionality of the hidden layer
    dim_output : int
        the dimensionality of the output layer
    Returns
    -------
    y_pred : float
        the output of the computation of the forward pass of the network
    """
    X_bb = np.c_[X, np.ones((X.shape[0], 1))] # bb = by bias

    W_1 = np.random.rand(dim_input + 1, dim_hidden + 1) # plus 1 for bias
    W_2 = np.random.rand(dim_hidden + 1, dim_output)

    y_pred = activation(X_bb.dot(W_1)).dot(W_2)
    return y_pred
    
```
